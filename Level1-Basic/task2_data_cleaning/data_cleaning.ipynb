```python
# 'pandas' lets us load and work with tabular data
import pandas as pd

# Load both churn files - remember, they're just an 80%/20% split of the same dataset
df_80 = pd.read_csv("churn-bigml-80.csv")
df_20 = pd.read_csv("churn-bigml-20.csv")

# Combine them into one single dataset using concat (short for "concatenate" = join together)
df = pd.concat([df_80, df_20], ignore_index=True)

# Quick confirmation that the combine worked
print("Combined dataset shape:", df.shape)
```

```python
# Question 1: How big is this dataset? (already know it's 3333 rows, 20 columns, but let's confirm)
print("Shape:", df.shape)

# Question 2: What columns exist, and what type of data is in each?
print("\nData types:")
print(df.dtypes)

# Question 3: What does the actual data look like?
print("\nFirst 5 rows:")
print(df.head())
```

```python
# Question 4: Is anything missing in any column?
print("Missing values per column:")
print(df.isnull().sum())

# Question 5: Are there any exact duplicate rows?
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

```python
# Get a statistical summary of all numeric columns at once
df.describe()
```

```python
# Step 1: Find the 25% and 75% marks (called Q1 and Q3) for this column
Q1 = df["Customer service calls"].quantile(0.25)
Q3 = df["Customer service calls"].quantile(0.75)

# Step 2: Calculate the IQR - the range where the "middle 50%" of the data sits
IQR = Q3 - Q1

# Step 3: Calculate the lower and upper "fences" - anything outside these is an outlier
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print("Q1 (25%):", Q1)
print("Q3 (75%):", Q3)

print("IQR:", IQR)

print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)

# Step 4: Count how many rows fall outside these fences (our outlier suspects)
outliers = df[(df["Customer service calls"] < lower_fence) | (df["Customer service calls"] > upper_fence)]
print("\nNumber of outlier rows found:", len(outliers))
```

```python
# Same 4 steps as before, but for "Total intl calls"
Q1 = df["Total intl calls"].quantile(0.25)
Q3 = df["Total intl calls"].quantile(0.75)
IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print("Q1 (25%):", Q1)
print("Q3 (75%):", Q3)
print("IQR:", IQR)
print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)

outliers_intl = df[(df["Total intl calls"] < lower_fence) | (df["Total intl calls"] > upper_fence)]
print("\nNumber of outlier rows found:", len(outliers_intl))
```

```python
# scikit-learn's LabelEncoder turns text categories into numbers
from sklearn.preprocessing import LabelEncoder

# Create one encoder we can reuse
le = LabelEncoder()

# Apply it to "International plan" - this replaces "Yes"/"No" with 1/0
df["International plan"] = le.fit_transform(df["International plan"])

# Apply it to "Voice mail plan" the same way
df["Voice mail plan"] = le.fit_transform(df["Voice mail plan"])

# Let's check the result
print(df[["International plan", "Voice mail plan"]].head(10))
```

```python
# Churn is True/False - let's convert it to 1/0 for consistency
df["Churn"] =df["Churn"].astype(int)

# Quick check
print(df["Churn"].head(10))
print("\nUnique values in Churn:", df["Churn"].unique())
```

```python
# pd.get_dummies() creates one new 0/1 column per category
df = pd.get_dummies(df, columns=["State"], prefix="State")

# Let's check the new shape - we expect many more columns now
print("New shape:", df.shape)

# Let's peek at some of the new State columns
print(df.filter(like="State_").head())
```

```python
# Check row 0 specifically for the State_KS column
print(df.loc[0, "State_KS"])

# Also confirm only ONE state column is True for this row (since a customer can only be in one state)
print(df.loc[0].filter(like="State_").sum())
```

```python
# These are the genuine numeric columns we want to standardize
# (excluding our encoded 0/1 columns and the Churn target)

numeric_cols = [ 
    "Account length", "Area code", "Number vmail messages",
    "Total day minutes", "Total day calls", "Total day charge",
    "Total eve minutes", "Total eve calls", "Total eve charge",
    "Total night minutes", "Total night calls", "Total night charge",
    "Total intl minutes", "Total intl calls", "Total intl charge",
    "Customer service calls"
]

print("number of columns to standerdize:", len(numeric_cols))
```

```python
# StandardScaler is scikit-learn's tool for standardization
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Fit it to our numeric columns, and transform them in one step
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

# Let's check the result
print(df[numeric_cols].head())
```

```python
# Save our fully cleaned and processed dataset to a new CSV file
df.to_csv("Churn_cleaned.csv", index=False)

print("Cleaned dataset saved as chun_cleaned.csv")
print("Final shape:", df.shape)
```
